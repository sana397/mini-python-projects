# Import Required Libraries
This section imports the necessary libraries for interactive input and display in the notebook.

```python
import asyncio
from functools import wraps
from typing import Any, Callable, TypeVar, Generic

# Advanced metaclass with descriptor protocol
class SingletonMeta(type):
    _instances = {}
    
    def __call__(cls, *args, **kwargs):
        if cls not in cls._instances:
            cls._instances[cls] = super().__call__(*args, **kwargs)
        return cls._instances[cls]

# Decorator for async retry logic
def async_retry(max_retries: int = 3, delay: float = 1.0):
    def decorator(func: Callable) -> Callable:
        @wraps(func)
        async def wrapper(*args, **kwargs):
            for attempt in range(max_retries):
                try:
                    return await func(*args, **kwargs)
                except Exception as e:
                    if attempt == max_retries - 1:
                        raise
                    await asyncio.sleep(delay)
        return wrapper
    return decorator

# Generic class with type parameters
T = TypeVar('T')

class AsyncCache(Generic[T], metaclass=SingletonMeta):
    def __init__(self):
        self._cache: dict[str, T] = {}
    
    async def get(self, key: str) -> T:
        return self._cache.get(key)
    
    async def set(self, key: str, value: T) -> None:
        self._cache[key] = value

# Complex async function with retry
@async_retry(max_retries=3, delay=0.5)
async def fetch_data(url: str) -> str:
    await asyncio.sleep(0.1)
    return f"Data from {url}"

async def main():
    cache = AsyncCache[str]()
    await cache.set("key1", "value1")
    result = await fetch_data("https://example.com")
    print(result)

import nest_asyncio
nest_asyncio.apply()
await main()
```
